## Read in olink data

```r
# from IDE for ALTRA Olink Data for Aim 1 (CON | ARI) 2024-10-22

olink_uuid <- '4bbd7e17-607f-4568-a7d2-8190a6eb0c59' # bokim-gomop-bukiv/2024-10-22_Olink_altraHCvARI.csv
```

```r
olink_file <- hise::cacheFiles(list(olink_uuid))

olink <- read.csv(olink_file, 
                  header = T, 
                  stringsAsFactors = F)

nrow(olink)
```

## Remove subject level metadata

```r
terms <- c('sample', 'subject', 'cohort', 'age', 'time')

keep_olink <- olink %>%
    select(SampleID,
           sample.sampleKitGuid,
           !contains(terms))

nrow(keep_olink)
```

```r
names(keep_olink)
```

#### Save filtered data

```r
out_olink <- paste0(
    "/home/workspace/output/",
    Sys.Date(),
    "_Olink_altraHCvARI_noMeta.csv"
)

write.csv(
    keep_olink, 
    out_olink,
    row.names = FALSE
)
```

## Store results in HISE

```r
study_space_uuid <- '223de760-9624-45bd-aefe-ca24c75b1800'

title <- paste('ALTRA Olink Data for Aim 1: CON | ARI (No Subject Metadata)', Sys.Date())
```

```r
search_id <- ids::proquint(n_words = 3)
search_id
```

```r
in_list <- list(olink_uuid)
```

```r
out_list <- list(out_olink)
```

```r
uploadFiles(
    files = out_list,
    studySpaceId = study_space_uuid,
    title = title,
    inputFileIds = in_list,
    destination = search_id,
    store = "project",
    doPrompt = FALSE
)
```
