## TmemC3 to B effector


```python
liana_res.loc[(liana_res['source'] == 'CD4mem_C3') &
              (liana_res['cellphone_pvals'] < 0.05)].head(10)
```

```python
# plot the magnitute from c3 to C9 and C8
c3_c8_c9_liana = liana_res.loc[(liana_res['source'] == 'CD4mem_C3') &
                               (liana_res['target'].isin(['CD27- effector B cell_C9', 'CD27- effector B cell_C8']))]
mag_c3_c8c9 = pd.pivot_table(c3_c8_c9_liana, index=['source', 'ligand_complex', 'receptor_complex'],
                             columns='target', values='magnitude_rank', fill_value=1).reset_index()

# get the index for the top 15 ligand receptor pair in both comparison
c3_c9_top = mag_c3_c8c9.sort_values(by='CD27- effector B cell_C9').head(25)
c3_c8_top = mag_c3_c8c9.sort_values(by='CD27- effector B cell_C8').head(25)

mag_c3_c8c9['ligand_receptor_complex'] = ""
mag_c3_c8c9.loc[c3_c9_top.index.append(c3_c8_top.index),
                'ligand_receptor_complex'] = mag_c3_c8c9.loc[c3_c9_top.index.append(c3_c8_top.index),
                                                             'ligand_complex'] + '_' + mag_c3_c8c9.loc[c3_c9_top.index.append(c3_c8_top.index), 'receptor_complex']
# only show top 30 labels
mag_c3_c8c9['ligand_receptor_complex']
mag_c3_c8c9.head()
```

```python
import plotly.express as px

fig = px.scatter(mag_c3_c8c9, x='CD27- effector B cell_C8',
                 y='CD27- effector B cell_C9', hover_data='ligand_receptor_complex',
                 text="ligand_receptor_complex")
fig.update_layout(
    xaxis=dict(autorange="reversed"),  # Reverse the x-axis
    yaxis=dict(autorange="reversed")   # Reverse the y-axis (if needed)
)
fig.update_layout(
    xaxis=dict(scaleanchor="y",  # Link the x-axis scale to the y-axis
               constrain="domain"),  # Ensures x-axis fits within the domain
    yaxis=dict(constrain="domain")  # Ensures y-axis fits within the domain
)
fig.update_traces(
    textposition="top center",  # Position labels above the points
    textfont=dict(size=6, color="black")  # Customize label font
)
fig.show()
```

```python
import matplotlib.pyplot as plt
from adjustText import adjust_text

# Example Data
x = mag_c3_c8c9['CD27- effector B cell_C8']
y = mag_c3_c8c9['CD27- effector B cell_C9']
labels = mag_c3_c8c9['ligand_receptor_complex']

# Create a Seaborn Scatter Plot
fig, ax = plt.subplots(figsize=(8, 8))
# Adjust `s` for marker size
sns.scatterplot(x=x, y=y, ax=ax, color="blue", s=50)

# Add Text Labels
texts = [ax.text(x[i], y[i], labels[i], fontsize=10) for i in range(len(x))]

# Adjust Text to Prevent Overlap
adjust_text(
    texts,
    # Optional arrows to indicate points
    arrowprops=dict(arrowstyle="->", color="gray", lw=0.5)
)
# Reverse x-axis and y-axis
ax.invert_xaxis()
ax.invert_yaxis()
# Customize Plot
ax.set_title("Cell-cell interation based on static expression", fontsize=14)
ax.set_xlabel("TmemC3_to_BeffC8")
ax.set_ylabel("TmemC3_to_BeffC9")

# Show Plot
plt.tight_layout()
plt.savefig(fig_path + "ALTRA_Tmem_c3_to_C8C9_liana_res.png",
            dpi=300, bbox_inches="tight")  # High-quality PNG

plt.show()
```
