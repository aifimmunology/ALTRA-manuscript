```python
import sys
import getopt
import matplotlib.pyplot as plt
import shutil
import h5py
import re
import os
import numpy as np
import pandas as pd
import scanpy as sc
import celltypist
import gc
import anndata
from celltypist import models
import h5py
import scipy.sparse as scs
import scrublet
from multiprocessing import Pool
```

```python
h5_list = ['prime3_lauren_align/B007-P2_PB00257-01_2020-06-25T12:38:29.677205289Z_labeled.h5',
           'prime3_lauren_align/B026-P1_PB00287-01_2020-10-10T20:09:23.161979427Z_labeled.h5',
           'prime3_lauren_align/B072-P1_PB01565-01_2021-07-12T17:24:36.579193615Z_labeled.h5',
           'prime3_lauren_align/B075-P1_PB01606-01_2021-07-19T01:12:47.537590436Z_labeled.h5']
```

```python
for h5_in in h5_list:
    print(h5_in)

    # Perform Scrublet & celltypist
    adata = sc.read_10x_h5(h5_in)
    sc.pp.normalize_total(adata, target_sum=1e4)
    sc.pp.log1p(adata)
    predictions_l1 = celltypist.annotate(adata, model='ref_pbmc_clean_celltypist_model_AIFI_L1_2024-04-18.pkl')
    df1 = predictions_l1.predicted_labels
    predictions_l2 = celltypist.annotate(adata, model='ref_pbmc_clean_celltypist_model_AIFI_L2_2024-04-19.pkl')
    df2 = predictions_l2.predicted_labels
    predictions_l3 = celltypist.annotate(adata, model='ref_pbmc_clean_celltypist_model_AIFI_L3_2024-04-19.pkl')
    df3 = predictions_l3.predicted_labels
    adata.shape
    scrub = scrublet.Scrublet(adata.X)
    doublet_scores, predicted_doublets = scrub.scrub_doublets(verbose = False)
    
    # Add Scrublet information to h5 file
    h5_con = h5py.File(h5_in, 'r+')
    h5_con['matrix']['observations']['AIFI_PBMC_L1'] = df1['predicted_labels'].tolist()
    h5_con['matrix']['observations']['AIFI_PBMC_L2'] = df2['predicted_labels'].tolist()
    h5_con['matrix']['observations']['AIFI_PBMC_L3'] = df3['predicted_labels'].tolist()
    h5_con['matrix']['observations']['scrublet_score'] = doublet_scores
    h5_con['matrix']['observations']['scrublet_call'] = predicted_doublets.astype(int)
    h5_con.close()
```
